**Q1:** Aik `Student` class banayein jismein `name` aur `roll_number` private variables (`__`) hon. In variables ko access aur modify karne ke liye getter aur setter methods likhein.

```python
# Write your code here for Q1
class Student:
    def __init__(self, name, roll_number):
        self.__name = name
        self.__roll_number = roll_number
        
    def setter(self, name, roll_number):
        self.__name = name
        self.__roll_number = roll_number

    def getter(self):
        return f"{self.__name} and {self.__roll_number}"

s1 = Student('amir',1)
s1.setter("azan",0)
s1.getter()
```

**Q2:** Aik `BankAccount` class banayein jismein `__balance` private variable ho. Aik `deposit()` aur aik `withdraw()` method banayein, check karein ke withdraw karte waqt balance zero se kam na ho.

```python
# Write your code here for Q2
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        self.__balance+= amount

    def withdraw(self, amount):
        if amount > self.__balance or self.__balance < 0:
            print("insufficient balance, withdraw failed")
        else:
            self.__balance -= amount
            print('withdraw sucessfully')
    def getBalance(self):
        print(f"your balance is : {self.__balance}")

ac1 = BankAccount(10000)
ac1.deposit(2300)
ac1.withdraw(100000)
ac1.getBalance()

```

**Q6:** Single Inheritance ka use karte hue aik parent class `Vehicle` (with brand attribute) aur child class `Car` (with model attribute) banayein. Car ka object bana kar dono attributes print karein.

```python
# Write your code here for Q6
class Vehicle:
    def __init__(self,brand):
        self.__brand = brand

    def getBrand(self):
        return self.__brand

class Car(Vehicle):
    def __init__(self, brand, model):
        self.__model = model
        super().__init__(brand)

    def getCar(self):
        return(f"Model: {self.__model} brand : {super().getBrand()}")

bmw = Car("BMW", "i series")
bmw.getCar()
```

**Q7:** Multilevel Inheritance ka use karein. Class `Animal` -> Class `Mammal` -> Class `Dog`. Dog class ke object se parent classes ke features ko access karein.

```python
# Write your code here for Q7
class Animal:
    def __init__(self,category):
        self.__category = category
    def show(self):
        return(f"category: {self.__category}")
        
class Mammal(Animal):
    def __init__(self,name,age,weight,category):
        super().__init__(category)
        self.__name = name
        self.__age= age
        self.__weight = weight

    def show(self):
        return(f"{ super().show()} name :{self.__name} age: {self.__age} weight {self.__weight} ")
    
class Dog(Mammal):
    def __init__ (self, category,name,age,weight,breed,color,is_trained):
        super().__init__(name ,age,weight,category)
        self.__breed = breed
        self.__color = color
        self.__is_trained = is_trained
    def show(self):
        print(f" {super().show()} breed: { self.__breed} color: {self.__color} is_trained: {self.__is_trained}")
D1 = Dog("Mamal","tommy",12,53,"german","white",True)
D1.show()
```

**Q9:** Child class ke constructor (`__init__`) ke andar parent class ke constructor ko call karne ke liye `super()` ka use kaise hota hai? Aik practical example likhein.

```python
# Write your code here for Q9
class Mother:
    def __init__ (self, skin_color):
        self.__skin_color = skin_color
    def show(self):
        return f"skin_color: {self.__skin_color}"
class Child( Mother):
    def __init__ (self,name,skin_color):
        self.__name  = name
        super().__init__(skin_color)
    def show(self):
        return f"name: {self.__name}, {super().show()}"

c = Child("Amir",'brown')
c.show()
```

**Q10:** Hierarchical Inheritance kya hoti hai? Aik parent class `Shape` banayein aur us se do alag child classes `Circle` aur `Square` banakar samjhayein.

```python
# Write your code here for Q10
# Hierarchical Inheritance ma ak parent se 2 ya 2 se ziada child inherit hote hai

class Shape:
    def __init__ (self,area, perimeter):
        self.__area = area
        self.__perimeter = perimeter
    def show(self):
        return f"A: {self.__area} P: {self.__perimeter}"
        
class Circle(Shape):
    def __init__ (self,radius,area, perimeter):
        self.__radius  = radius
        super().__init__(area, perimeter)
    def show(self):
        return f"radius: {self.__radius}, {super().show()}"

class Square(Shape):
    def __init__ (self,length,width,area, perimeter):
        self.__length  = length
        self.__width = width
        super().__init__(area, perimeter)
    def show(self):
        return f"name: L = {self.__length}, W = {self.__width} {super().show()}"    

s1 = Square(12,34,2.3,2.33)
s1.show()
    
```

**Q11:** Method Overriding kya hai? Parent class `Bird` banayein jisme `fly()` method ho jo print kare 'Most birds can fly'. Child class `Ostrich` banayein jo is method ko override kare aur print kare 'Ostriches cannot fly'.

```python
# Write your code here for Q11
# overriding means changing the function orignality
class Bird:
    def fly(self):
        print('most birds can fly')
class ostrich(Bird):
    def fly(self):
        print("ostrich cannot fly")
b = Bird()
o = ostrich()
b.fly()
o.fly()
    
```

**Q12:** Polymorphism function ke zariye kaise achieve hoti hai? Aik function banayein `make_sound(animal_object)` jo har tarah ke animal class ka sound method automatically call kare.

```python
# Write your code here for Q12

class Bird:
    def sound(self):
        print("Bird makes a sound")


class Sparrow(Bird):
    def sound(self):
        print("Sparrow chirps")


class Duck(Bird):
    def sound(self):
        print("Duck quacks")


b = Bird()
s = Sparrow()
d = Duck()

b.sound()
s.sound()
d.sound()
```

**Q13:** Operator Overloading kya hoti hai? Do objects ko `+` operator ke zariye add karne ke liye `__add__` magic method ka use kaise karte hain? Code likhein.

```python
# Write your code here for Q13
# operator overloading means operator k behaviour ko change krna
class Number:
    def __init__(self, num):
        self.__num =  num
    def __add__(self, other):
        return self.__num + other.__num

n1 = Number(2)
n2 = Number(4)
print(n1 + n2)
```

**Q16:** `abc` module ka use karte hue aik abstract class `Computer` banayein jisme abstract method `process()` ho. Is class ka object banane ki koshish karein aur dekhein kya hota hai.

```python
# Write your code here for Q16
from abc import ABC, abstractmethod
class Computer(ABC):
    @abstractmethod
    def process(self):
        print('i am the process method')
a = Computer()
print(a.process())
```

**Q17:** Q16 wali `Computer` class ko inherit karke do child classes `Laptop` aur `Desktop` banayein aur dono mein `process()` method ko mandatory implement karein.

```python
# Write your code here for Q17
from abc import ABC, abstractmethod
class Computer(ABC):
    @abstractmethod
    def process(self):
        print('i am the computerrprocess method')
class Laptop(Computer):
    def process(self):
        print('i am the laptop processor')
class Desktop(Computer):
    def process(self):
        print('i am the Desktop processor')
l = Laptop()
d = Desktop()

d.process()
l.process()
```

**Q18:** Aik abstract class `Car` banayein jisme aik concrete (normal) method `start_engine()` ho aur aik abstract method `drive()` ho. Child class bana kar dono ko test karein.

```python
# Write your code here for Q18
from abc import ABC, abstractmethod
class Car(ABC):
    @abstractmethod
    def drive(self):
        print('car is driving')
    def start_engine(self):
        print('the engine start')

class ElectricCar(Car):
    def drive(self):
        print('Electric car is driving')

ev = ElectricCar()
ev.drive()
ev.start_engine()
```

**Q20:** Aik real-world scenario (jaise Payment System) design karein jahan `Payment` aik abstract class ho aur `CreditCardPayment` aur `UPIPayment` uski subclasses hon.

```python
# Write your code here for Q20
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self):
        print('payment!')

class CreditCardPayment(Payment):
    def pay(self):
        print('CreditCardPayment ')
        
class UPIPayment(Payment):
    def pay(self):
        print('UPI Payment ')
c = CreditCardPayment()
u = UPIPayment()
c.pay()
u.pay()
```
