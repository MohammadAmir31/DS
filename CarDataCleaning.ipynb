```python
valid_company_name = ['Hyundai', 'Mahindra', 'Maruti', 'Ford', 'Skoda', 'Audi', 'Toyota',
                      'Renault', 'Honda', 'Datsun', 'Mitsubishi', 'Tata', 'Volkswagen','Chevrolet',
                      'Mini', 'BMW', 'Nissan', 'Hindustan', 'Fiat','MARUTI', 'Force', 'Mercedes', 'Land', 'Yamaha',
                        'Swift','Jaguar', 'Volvo','TATA']
```

```python
data.company = data.company.apply(lambda x: x if x in valid_company_name else 'Maruti')
```

```python
data.year.unique()
```

```python
d1 = data[data.year.str.isnumeric()]
d2 = data[~(data.year.str.isnumeric())]
```

```python
d1.year = d1.year.astype(int)
```

```python
d2.year = d1.year.mode()[0]
```

```python
new_data = pd.concat([d1,d2])
```

```python
new_data.info()
```

```python
new_data.Price.unique()
```

```python
d1 = new_data[new_data.Price=='Ask For Price']
d2 = new_data[new_data.Price!='Ask For Price']
```

```python
d2.Price = d2.Price.str.replace(',','').astype(int)
```

```python
d1.Price = d1.Price.replace('Ask For Price',d2.Price.median())
```

```python
new_data = pd.concat([d1,d2])
```

```python
new_data.info()
```

```python
new_data.kms_driven.unique()
```

```python
d1 = new_data[new_data.kms_driven=='Petrol']
d2  = new_data[new_data.kms_driven!='Petrol']
```

```python
d2.kms_driven = d2.kms_driven.str.replace(',', '').str.replace('kms', '').astype(float)
```

```python
d2.kms_driven.median()
```

```python
d1.kms_driven = d1.kms_driven.replace('Petrol', 41000.0)
```

```python
new_data = pd.concat([d1,d2])
```
